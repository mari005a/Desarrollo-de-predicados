```python
# Universo animales
universo = {
    "mamiferos": {
        "Elefante": ["TienePelo", "DaLeche", "TieneTrompa", "OrejasGrandes", "GranAltura"],
        "Guepardo": ["TienePelo", "ComeCarne", "ColorLeonado", "ManchasOscuras"],
        "Tigre": ["TienePelo", "ComeCarne", "ColorLeonado", "RayasNegras"],
        "Jirafa": ["DaLeche", "TienePezunas", "CuelloLargo", "ManchasOscuras"],
        "Cebra": ["TienePelo", "TienePezunas", "RayasNegras"]
    },

    "aves": {
        "Pinguino": ["TienePlumas", "Nada", "BlancoNegro"],
        "Albatros": ["TienePlumas", "Vuela", "PoneHuevos", "VuelaMuyBien"]
    }
}

# Tuplas / Hechos
hechos = {

    ("Elefante", "tiene", "Pelo"),
    ("Elefante", "da", "Leche"),
    ("Elefante", "tiene", "Trompa"),
    ("Elefante", "tiene", "OrejasGrandes"),
    ("Elefante", "tiene", "GranAltura"),

    ("Guepardo", "tiene", "Pelo"),
    ("Guepardo", "come", "Carne"),
    ("Guepardo", "tiene", "ColorLeonado"),
    ("Guepardo", "tiene", "ManchasOscuras"),

    ("Tigre", "tiene", "Pelo"),
    ("Tigre", "come", "Carne"),
    ("Tigre", "tiene", "ColorLeonado"),
    ("Tigre", "tiene", "RayasNegras"),

    ("Jirafa", "da", "Leche"),
    ("Jirafa", "tiene", "Pezunas"),
    ("Jirafa", "tiene", "CuelloLargo"),
    ("Jirafa", "tiene", "ManchasOscuras"),

    ("Cebra", "tiene", "Pelo"),
    ("Cebra", "tiene", "Pezunas"),
    ("Cebra", "tiene", "RayasNegras"),

    ("Pinguino", "tiene", "Plumas"),
    ("Pinguino", "nada", "Si"),
    ("Pinguino", "tiene", "BlancoNegro"),

    ("Albatros", "tiene", "Plumas"),
    ("Albatros", "vuela", "MuyBien"),
    ("Albatros", "pone", "Huevos")
}


# Auxiliar: caracteristicas de un animal
def caracteristicas(x):
    for grupo in universo:
        if x in universo[grupo]:
            return universo[grupo][x]
    return []


# Predicados
def mamifero(x):
    return x in universo["mamiferos"]

def ave(x):
    return x in universo["aves"]

def carnivoro(x):
    return mamifero(x) and "ComeCarne" in caracteristicas(x)

def ungulado(x):
    return mamifero(x) and "TienePezunas" in caracteristicas(x)

def guepardo(x):
    c = caracteristicas(x)
    return carnivoro(x) and "ColorLeonado" in c and "ManchasOscuras" in c

def tigre(x):
    c = caracteristicas(x)
    return carnivoro(x) and "ColorLeonado" in c and "RayasNegras" in c

def jirafa(x):
    c = caracteristicas(x)
    return ungulado(x) and "CuelloLargo" in c and "ManchasOscuras" in c

def cebra(x):
    return ungulado(x) and "RayasNegras" in caracteristicas(x)

def pinguino(x):
    c = caracteristicas(x)
    return ave(x) and "Nada" in c and "BlancoNegro" in c

def albatros(x):
    return ave(x) and "VuelaMuyBien" in caracteristicas(x)

def elefante(x):
    c = caracteristicas(x)
    regla13 = mamifero(x) and "TieneTrompa" in c
    regla14 = "TieneTrompa" in c and "OrejasGrandes" in c and "GranAltura" in c
    return regla13 or regla14


# Consultas
print("1. Mamifero(Guepardo):", mamifero("Guepardo"))
print("2. Mamifero(Pinguino):", mamifero("Pinguino"))

print("3. Ave(Pinguino):", ave("Pinguino"))
print("4. Ave(Jirafa):", ave("Jirafa"))

print("5. Carnivoro(Tigre):", carnivoro("Tigre"))
print("6. Carnivoro(Cebra):", carnivoro("Cebra"))

print("7. Ungulado(Jirafa):", ungulado("Jirafa"))
print("8. Ungulado(Pinguino):", ungulado("Pinguino"))

print("9. Guepardo(Guepardo):", guepardo("Guepardo"))
print("10. Guepardo(Tigre):", guepardo("Tigre"))

print("11. Tigre(Tigre):", tigre("Tigre"))
print("12. Tigre(Cebra):", tigre("Cebra"))

print("13. Jirafa(Jirafa):", jirafa("Jirafa"))
print("14. Jirafa(Cebra):", jirafa("Cebra"))

print("15. Cebra(Cebra):", cebra("Cebra"))
print("16. Cebra(Albatros):", cebra("Albatros"))

print("17. Pinguino(Pinguino):", pinguino("Pinguino"))
print("18. Pinguino(Albatros):", pinguino("Albatros"))

print("19. Albatros(Albatros):", albatros("Albatros"))
print("20. Albatros(Pinguino):", albatros("Pinguino"))

print("21. Elefante(Elefante):", elefante("Elefante"))
print("22. Elefante(Tigre):", elefante("Tigre"))
```
